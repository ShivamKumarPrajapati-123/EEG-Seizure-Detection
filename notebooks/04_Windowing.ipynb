```python
import mne
import numpy as np
```

```python
file_path = r"C:\Users\Prajapati_Shivam\EEG-Seizure-Detection\data\chb01\chb01_01.edf"
raw = mne.io.read_raw_edf(file_path, preload= True)
print(raw)
```

```python
raw.filter(l_freq=0.5, h_freq=40)
```

```python
sfreq = int(raw.info["sfreq"])
print("Sampling frequency =", sfreq)
```

```python
window_seconds = 4
samples_per_window = sfreq*window_seconds
print("Samples per Window =", samples_per_window)
```

```python
data = raw.get_data()
print(data.shape)
```

```python
total_samples = data.shape[1]
print("Total Samples =", total_samples)
```

```python
num_windows = total_samples // samples_per_window
print("Numbers of Windows =", num_windows)
```

```python
windows = []

for i in range(num_windows):
    start = i * samples_per_window
    end = start + samples_per_window
    window = data[:, start:end]
    windows.append(window)
print("Total Windows Created =", len(windows))
```

```python
print(windows[0].shape)
```

```python
import numpy as np
windows = np.array(windows)
print(windows.shape)
```

```python
np.save("../data/windows.npy",windows)
print("Windows saved sucessfully!")
```
