```python
import os 
data_folder = "../data/chb01"

edf_files = sorted([
    file for file in os.listdir(data_folder)
    if file.endswith(".edf")
])
print("Number of EDF files:", len(edf_files))
```

```python
for file in edf_files:
    print(file)
```

```python
import mne 
import os
for file in edf_files:
    file_path = os.path.join(data_folder, file)
    raw = mne.io.read_raw_edf(file_path, preload=False, verbose=False)
    print(file, "->", raw.info["sfreq"], "Hz")
```

```python
import mne 
import os
for file in edf_files:
    file_path = os.path.join(data_folder, file)
    raw = mne.io.read_raw_edf(
        file_path,
        preload=False,
        verbose=False
    )
    duration = raw.n_times / raw.info["sfreq"]
    print(file, "->", duration, "seconds")
    
```

```python
seizure_start = 2996
seizure_end = 3036

print("Seizure Start:", seizure_start, "seconds")
print("Seizure End:", seizure_end, "seconds")
```

```python
labels = np.zeros(len(windows), dtype=int)
print("Total Windows:", len(labels))
```

```python
window_duration = 4
for i in range(len(windows)):
    window_start = i*window_duration
    window_end = window_start + window_duration

    if window_start < seizure_end and window_end > seizure_start:
        labels[i] = 1
```

```python
print("Total Windows :", len(labels))
print("Seizure Windows :", np.sum(labels))
print("Normal Windows :", len(labels) - np.sum(labels))
```

```python
seizure_indices = np.where(labels == 1)[0]
print("Seizure Windows Indices:")
print(seizure_indices)
```

```python
for i in seizure_indices:
    start = i*window_duration
    end = start + window_duration

    print(
    f"Window {i}:"
    f"{start}-{end} seconds"
    )
```

```python
print("Last seizure window:")
i = seizure_indices[-1]
start = i*window_duration
end = start + window_duration
print(F"Window {i}: {start}-{end} seconds")
```

```python
np.save("../data/labels.npy", labels)
print("Labels saved successfully!")
```
